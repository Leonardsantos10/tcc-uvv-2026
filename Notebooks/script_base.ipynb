```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Datas
datas = pd.date_range(start="2021-01-01", end="2025-12-31", freq="D")

# Empresas
empresas = [f"EMP{i}" for i in range(1, 81)]

# Produtos
produtos = ["Parabrisa", "Vigia", "Laterais", "Retrovisor", "Farol"]

# Volume médio por produto
base_volume = {
    "Parabrisa": 30,
    "Vigia": 25,
    "Laterais": 22,
    "Farol": 15,
    "Retrovisor": 10
}

def fator_sazonal(data):
    mes = data.month

    # Alta no começo do ano
    if mes in [1,2,3]:
        return 1.3

    # Meio do ano estável e baixo
    elif mes in [4,5,6,7,8,9]:
        return 0.8

    # Alta final do ano
    else:
        return 1.4

# Crescimento anual de ~5% a partir de 2021
def fator_crescimento(data):
    anos_decorridos = data.year - 2021
    return 1.05 ** anos_decorridos

dados = []

for data in datas:
    fator = fator_sazonal(data)
    crescimento = fator_crescimento(data)

    for emp in empresas:
        for prod in produtos:

            media = base_volume[prod] * fator * crescimento

            volume = np.random.poisson(media)

            dados.append([
                data,
                emp,
                prod,
                volume
            ])

df = pd.DataFrame(
    dados,
    columns=[
        "data",
        "empresa",
        "produto",
        "volume_atendimentos"
    ]
)

print(df.head())
print(len(df))
```

```python
df.to_csv('VolumeDeAtendimentos.csv', sep=';')
```
