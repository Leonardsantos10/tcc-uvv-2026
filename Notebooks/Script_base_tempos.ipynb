```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Datas
datas = pd.date_range(start="2021-01-01", end="2025-12-31", freq="D")

# Empresas (EMP01 até EMP81)
empresas = [f"EMP{i:02d}" for i in range(1, 82)]

# Produtos
produtos = ["Parabrisa", "Vigia", "Laterais", "Farol", "Retrovisor"]

# Tempo médio base por produto (em minutos)
tempo_base = {
    "Parabrisa":  45,
    "Vigia":      65,
    "Laterais":   55,
    "Farol":      35,
    "Retrovisor": 30
}

# Fator de velocidade por empresa:
# Valores < 1.0 → empresa mais rápida | Valores > 1.0 → empresa mais lenta
# Distribuição uniforme entre 0.85 e 1.15 (~±15% sobre o tempo base)
fator_empresa = {emp: np.random.uniform(0.85, 1.15) for emp in empresas}

dados = []

for data in datas:
    for emp in empresas:
        for prod in produtos:

            # Tempo esperado para essa empresa/produto
            media = tempo_base[prod] * fator_empresa[emp]

            # Variação diária aleatória (desvio padrão ~7 min → cobre ~±20 min)
            tempo = np.random.normal(media, 7)

            # Garante tempo mínimo razoável
            tempo = max(10.0, round(tempo, 1))

            dados.append([
                data,
                emp,
                prod,
                tempo
            ])

df = pd.DataFrame(
    dados,
    columns=[
        "data",
        "empresa",
        "produto",
        "tempo_medio_min"
    ]
)

print(df.head(10))
print(f"
Total de registros: {len(df):,}")
print("
Tempo médio por produto:")
print(df.groupby("produto")["tempo_medio_min"].mean().round(1))
```

```python
# Verificação: empresas mais rápidas vs mais lentas
ranking = pd.DataFrame({
    "empresa": list(fator_empresa.keys()),
    "fator_velocidade": list(fator_empresa.values())
}).sort_values("fator_velocidade")

print("5 empresas mais rápidas (menor fator):")
print(ranking.head())
print("
5 empresas mais lentas (maior fator):")
print(ranking.tail())
```

```python
df.to_csv('TempoMedioAtendimentos.csv', sep=';', index=False)
```
