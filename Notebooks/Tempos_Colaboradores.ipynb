```python
import pandas as pd
import numpy as np

np.random.seed(42)
rng = np.random.default_rng(42)

nomes_primeiro = [
    'Andre', 'Bruno', 'Carlos', 'Daniel', 'Eduardo', 'Fernando', 'Gabriel',
    'Henrique', 'Igor', 'Joao', 'Leonardo', 'Marcos', 'Nicolas', 'Pedro',
    'Rafael', 'Rodrigo', 'Samuel', 'Thiago', 'Victor', 'Wilson',
    'Ana', 'Beatriz', 'Carolina', 'Diana', 'Elisa', 'Fernanda', 'Gabriela',
    'Helena', 'Isabela', 'Juliana', 'Karen', 'Larissa', 'Marina', 'Natalia',
    'Olivia', 'Patricia', 'Raquel', 'Sandra', 'Tatiana', 'Vanessa'
]

nomes_sobrenome = [
    'Silva', 'Santos', 'Oliveira', 'Souza', 'Lima', 'Pereira', 'Alves',
    'Rodrigues', 'Costa', 'Martins', 'Gomes', 'Ribeiro', 'Carvalho',
    'Ferreira', 'Almeida', 'Dias', 'Barbosa', 'Nascimento', 'Castro',
    'Monteiro', 'Pinto', 'Ramos', 'Moreira', 'Mendes', 'Rocha',
    'Nunes', 'Cunha', 'Teixeira', 'Coelho', 'Cardoso'
]

# 40 primeiros x 30 sobrenomes = 1200 combinacoes unicas
todas_combinacoes = [f'{p} {s}' for p in nomes_primeiro for s in nomes_sobrenome]

def gerar_pool_nomes(empresa, n):
    emp_num = int(empresa[3:])  # EMP07 -> 7
    local_rng = np.random.default_rng(emp_num * 997)
    indices = local_rng.permutation(len(todas_combinacoes))[:n]
    return [todas_combinacoes[i] for i in indices]
```

```python
df_colab = pd.read_csv('../Bases/ColaboradoresPorDia.csv', sep=';')
df_tempo = pd.read_csv('../Bases/TempoMedioAtendimentos.csv', sep=';')

print('ColaboradoresPorDia:', df_colab.shape)
print('TempoMedioAtendimentos:', df_tempo.shape)
```

```python
# Normaliza EMP1 -> EMP01, EMP10 -> EMP10, etc.
df_colab['empresa'] = (
    df_colab['empresa']
    .str.extract(r'EMP(\d+)')[0]
    .astype(int)
    .apply(lambda x: f'EMP{x:02d}')
)

# EMP81 esta em TempoMedioAtendimentos mas nao em ColaboradoresPorDia
datas_unicas = sorted(df_colab['data'].unique())
emp81_rng = np.random.default_rng(81 * 997)
df_emp81 = pd.DataFrame({
    'data': datas_unicas,
    'empresa': 'EMP81',
    'quantidade_colaboradores': emp81_rng.integers(5, 13, size=len(datas_unicas))
})
df_colab = pd.concat([df_colab, df_emp81], ignore_index=True)

print('Total linhas ColaboradoresPorDia (com EMP81):', len(df_colab))
```

```python
# Pool fixo de nomes por empresa (tamanho = max colaboradores da empresa)
max_colab_por_empresa = df_colab.groupby('empresa')['quantidade_colaboradores'].max()

lookup_records = []
for empresa, max_n in max_colab_por_empresa.items():
    nomes = gerar_pool_nomes(empresa, int(max_n))
    for idx, nome in enumerate(nomes):
        lookup_records.append({'empresa': empresa, 'colaborador_idx': idx, 'colaborador': nome})

df_nomes = pd.DataFrame(lookup_records)
print('Lookup criado:', len(df_nomes), 'registros')
print(df_nomes[df_nomes['empresa'] == 'EMP01'])
```

```python
# Join: tempo medio + quantidade de colaboradores por dia
df = df_tempo.merge(df_colab, on=['data', 'empresa'], how='left')
print('NaN em quantidade_colaboradores:', df['quantidade_colaboradores'].isna().sum())
print('Shape apos merge:', df.shape)
```

```python
# Expande cada linha (data, empresa, produto) em N linhas — uma por colaborador
df['quantidade_colaboradores'] = df['quantidade_colaboradores'].astype(int)

df_expandido = df.loc[df.index.repeat(df['quantidade_colaboradores'])].copy()
df_expandido = df_expandido.reset_index(drop=True)

# Indice 0..N-1 de cada colaborador dentro do grupo
df_expandido['colaborador_idx'] = df_expandido.groupby(['data', 'empresa', 'produto']).cumcount()

print('Shape expandido:', df_expandido.shape)
```

```python
# Associa nome via lookup
df_expandido = df_expandido.merge(df_nomes, on=['empresa', 'colaborador_idx'], how='left')

print('NaN em colaborador:', df_expandido['colaborador'].isna().sum())
print(df_expandido[['data', 'empresa', 'produto', 'colaborador', 'tempo_medio_min']].head(15))
```

```python
# Tempo individual: variacao +-15% em torno do tempo medio da empresa
ruido = rng.normal(loc=0.0, scale=0.15, size=len(df_expandido))
df_expandido['tempo_medio_min'] = (
    df_expandido['tempo_medio_min'] * (1.0 + ruido)
).clip(lower=1.0).round(1)

print(df_expandido['tempo_medio_min'].describe())
```

```python
# Base final
df_final = (
    df_expandido[['data', 'empresa', 'produto', 'colaborador', 'tempo_medio_min']]
    .sort_values(['data', 'empresa', 'produto', 'colaborador'])
    .reset_index(drop=True)
)

print('Shape final:', df_final.shape)
print(df_final.head(30))
```

```python
# Validacao: media reconstituida por empresa/produto/dia vs. TempoMedioAtendimentos
media_grupo = (
    df_final
    .groupby(['data', 'empresa', 'produto'])['tempo_medio_min']
    .mean()
    .reset_index()
    .rename(columns={'tempo_medio_min': 'media_calculada'})
)

validacao = media_grupo.merge(
    df_tempo.rename(columns={'tempo_medio_min': 'media_original'}),
    on=['data', 'empresa', 'produto']
)
validacao['diferenca_pct'] = (
    (validacao['media_calculada'] - validacao['media_original']).abs()
    / validacao['media_original'] * 100
)

print(f'Diferenca percentual media: {validacao["diferenca_pct"].mean():.2f}%')
print(f'Diferenca percentual maxima: {validacao["diferenca_pct"].max():.2f}%')
```

```python
df_final.to_csv('../Bases/TempoColaboradores.csv', sep=';', index=False)
print('Arquivo salvo: ../Bases/TempoColaboradores.csv')
print('Total de registros:', len(df_final))
```
